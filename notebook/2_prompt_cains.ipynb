### Prompt Templates - Dynamic Prompts

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python

llm =ChatGoogleGenerativeAI(model="gemini-3.5-flash")
out = StrOutputParser()
```

```python
def transform_case(result:str):
    return result.upper()
```

```python
prompts =ChatPromptTemplate.from_messages([
    {"role":"system", "content":"You are a translator and translate input in {language} ?"},
    {"role":"user",  "content":"{query}"},
])
```

```python
final_promts= prompts.format_messages(language="Higlish ?" , query = "I love Python, and JavaScript ?")
```

```python
res = llm.invoke(final_promts)
res.content
```

```python
res = llm.invoke(prompts.format_messages(language="Higlish ?" , query = "I love Python, and JavaScript ?"))
res.content
```

```python
chains = prompts | llm | out | transform_case
```

```python
final_result = chains.invoke({"language":"Hinglish", "query":"I love LangChain ?"})
```
