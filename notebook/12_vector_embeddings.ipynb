```python
from dotenv import load_dotenv
load_dotenv()
```

```python
documents = [
    "Machine learning is a subset of artificial intelligence",
    "Deep learning uses neural networks with multiple layers",
    "Python is a programming language popular for data science",
    "India won the cricket world cup in 2011",
    "AI models can understand and generate human language"
]
```

```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings
```

```python
embeddings = GoogleGenerativeAIEmbeddings(model="gemini-embedding-2-preview" ,  output_dimensionality=768,)
```

```python
vector = embeddings.aembed_query(documents)
```

```python
from langchain_chroma import Chroma

vector_store = Chroma(
    collection_name="vector_collection",
    embedding_function=embeddings,
    persist_directory="./chroma_langchain_db",  
)
```

```python
result=vector_store.similarity_search(query="Data " , k=2)
```
